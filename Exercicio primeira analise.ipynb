Estatistica Básica 4° modulo

###Importando a biblioteca numpy para calculos matemáticos de dados

```python
import numpy as np
```

```python
lista_idades = [26,30,32,22,26,35,40,20,43,31,23,100]
```

```python
##calculando a média da lista

##Primeiro soma os itens da lista
np.sum(lista_idades)
```

```python
##Agora sabendo quantos numeros tem na lista
len(lista_idades)
```

```python
##Dividindo a soma dos numeros pela quantidades de items para obter a média
np.sum(lista_idades)/len(lista_idades)
```

```python
##Outra maneira de calcular a média usando uma função do numpy
media = np.mean(lista_idades)
media
```

```python
##Só usando o print para monstrar informações
print("A média da lista foi:",media)
```

```python
#Ordenando a lista para conseguir manipular a mediana
lista_idades.sort()
```

```python
#uma maneira de ordenar a lista e calcular a mediana de cabeça
lista_idades
```

```python
#usando a função própria do numpy que calcula a mediana
mediana = np.median(lista_idades)
mediana
```

```python
#Calculando o desvio padrão da coluna dados
#Desvio padrão quer dizer o quanto os numeros estão distantes ou perto da média
#Quanto mais perto de 0, mais perto da media(homogenio), quantos mais longe mais (heterogenio)
dados['IDADE'].std()
```
